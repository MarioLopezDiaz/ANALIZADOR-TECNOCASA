# 🧪 Batería de Experimentos Finales: SMOTE Inteligente, Estudio de Ablación y Modelo Híbrido (XGBoost + KNN)

## 🎯 Objetivo del Cuaderno
Este *notebook* agrupa la última fase de experimentación algorítmica del proyecto. Su objetivo es explorar técnicas avanzadas para forzar los límites predictivos del modelo, abordando la escasez de datos, demostrando empíricamente el valor de las variables de IA y aplicando correcciones de sesgo local en las predicciones.

## ⚙️ Metodología y Flujo de Trabajo

El cuaderno se divide en tres grandes experimentos arquitectónicos:

### 1. Data Augmentation Basado en Vecindad (SMOTE Customizado)
En lugar de introducir ruido aleatorio global, se implementa una función `generar_datos_sinteticos_inteligentes` inspirada en la técnica SMOTE (*Synthetic Minority Over-sampling Technique*). 
* Utiliza **K-Nearest Neighbors (KNN)** para encontrar pisos similares en el espacio de características.
* Genera nuevas muestras interpolando matemáticamente las variables continuas entre un piso base y su vecino, y heredando las variables categóricas.
* **Aislamiento Estricto:** La partición del *Test Set* se realiza de forma blindada *antes* de la generación sintética para garantizar que la evaluación se realice única y exclusivamente sobre datos reales, evitando cualquier fuga de datos (*Data Leakage*).

### 2. Estudio de Ablación (Feature Importance Empírica)
Para cuantificar el valor real que la Inteligencia Artificial aporta al modelo frente a la tasación tradicional, se define un torneo entre tres configuraciones:
* **SMALL (Esencial):** Simula un tasador clásico usando solo variables estructurales (superficie, baños, ascensor, planta, año).
* **MEDIUM (Smart):** Añade la Renta Neta Media y las variables clave de IA (nivel de lujo, luz).
* **LARGE (Completo):** Utiliza la totalidad del *dataset* procesado.

### 3. Modelo Híbrido Customizado (XGBoost + Corrección Espacial KNN)
Se diseña e implementa una clase personalizada `HybridValuator` (heredando de `BaseEstimator` y `RegressorMixin` de scikit-learn) para atacar específicamente el error en las colas de la distribución.
* **Fase 1 (Base):** Entrena un modelo XGBoost robusto.
* **Fase 2 (Auditoría de Sesgo):** El modelo predice sobre sus propios datos de entrenamiento y calcula una ratio de error local (identificando zonas o tipos de pisos donde tiende a sobreestimar o subestimar).
* **Fase 3 (Corrección Local):** Ante una nueva predicción, un algoritmo KNN busca los pisos históricos más similares. Se calcula la mediana del error de esos vecinos y se aplica un factor de corrección suavizado a la predicción original del XGBoost.

## 📊 Evaluación
Al igual que en *notebooks* anteriores, todos los experimentos devuelven sus métricas en espacio real (Euros) mediante la transformación exponencial inversa, y desglosan el MAE y el MAPE por **deciles de precio** para auditar el comportamiento del algoritmo en los valores extremos del mercado inmobiliario.

### **VERSION SMALL - MEDIUM - LARGE + DATA AUGMENTATION**

```python
import pandas as pd
import numpy as np
import warnings
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.neighbors import NearestNeighbors
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder, TargetEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.metrics import mean_absolute_error
from xgboost import XGBRegressor

# Configuración
warnings.filterwarnings('ignore')
SEED = 42
np.random.seed(SEED)

# ==============================================================================
# 1. FUNCIÓN DE AUMENTO DE DATOS (SMOTE INTELIGENTE)
# ==============================================================================
def generar_datos_sinteticos_inteligentes(df_original, n_samples_objetivo=5000, k_neighbors=5):
    """
    Genera datos sintéticos interpolando entre vecinos del conjunto de entrenamiento.
    """
    # Separamos numéricas y categóricas
    num_cols = df_original.select_dtypes(include=['number']).columns.tolist()
    cat_cols = df_original.select_dtypes(include=['object', 'category', 'bool']).columns.tolist()
    
    # Imputar para cálculo de distancias
    imputer = SimpleImputer(strategy='median')
    X_num = pd.DataFrame(imputer.fit_transform(df_original[num_cols]), columns=num_cols)
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_num)
    
    # KNN
    nn = NearestNeighbors(n_neighbors=k_neighbors)
    nn.fit(X_scaled)
    
    new_rows = []
    n_generados = 0
    
    # Si pedimos muchas muestras, optimizamos el bucle
    while n_generados < n_samples_objetivo:
        # Elegir base y vecino
        idx_base = np.random.randint(0, len(df_original))
        distances, indices = nn.kneighbors(X_scaled[idx_base].reshape(1, -1))
        idx_neighbor = np.random.choice(indices[0][1:])
        
        # Interpolar Numéricas
        alpha = np.random.random()
        row_base_num = X_num.iloc[idx_base]
        row_neighbor_num = X_num.iloc[idx_neighbor]
        new_row_num = row_base_num + alpha * (row_neighbor_num - row_base_num)
        
        # Copiar Categóricas (del vecino más cercano o base)
        if np.random.random() > 0.5:
            new_row_cat = df_original.iloc[idx_base][cat_cols].reset_index(drop=True)
        else:
            new_row_cat = df_original.iloc[idx_neighbor][cat_cols].reset_index(drop=True)
            
        # Unir
        combined_row = pd.concat([new_row_num, new_row_cat], axis=0)
        
        # Ruido leve en precio y superficie para evitar duplicados exactos
        for col in ['Precio', 'Superficie']:
            if col in combined_row:
                noise = np.random.normal(0, 0.01)
                combined_row[col] = combined_row[col] * (1 + noise)
                
        new_rows.append(combined_row)
        n_generados += 1
        
    df_synthetic = pd.DataFrame(new_rows)
    return df_synthetic

# ==============================================================================
# 2. CARGA Y PREPARACIÓN (SPLIT CORRECTO)
# ==============================================================================
ruta_parquet = "C:/Users/mario/OneDrive/Mario/CARRERA/TFG/ANALIZADOR-TECNOCASA/DATOS/tecnocasa_modelo.parquet"
df = pd.read_parquet(ruta_parquet, engine='fastparquet')

# Filtro Outliers
q_low, q_high = df['Precio'].quantile(0.01), df['Precio'].quantile(0.99)
df = df[(df['Precio'] > q_low) & (df['Precio'] < q_high)].copy()

print(f"📉 Datos Reales Totales: {len(df)}")

# --- PASO CRÍTICO: SPLIT ANTES DE NADA ---
# Reservamos el 20% de datos reales para el examen final. NADIE toca estos datos.
df_train_real, df_test_real = train_test_split(df, test_size=0.2, random_state=SEED)

print(f"🔒 Test Set Blindado: {len(df_test_real)} muestras (Solo reales)")
print(f"🔨 Train Set Base: {len(df_train_real)} muestras (Se usarán para generar sintéticos)")

# --- AUMENTO DE DATOS (SOLO EN TRAIN) ---
# Generamos, por ejemplo, 20.000 muestras sintéticas basadas en las 640 reales de train
print("🧬 Generando datos sintéticos sobre el Train Set...")
df_sintetico = generar_datos_sinteticos_inteligentes(df_train_real, n_samples_objetivo=20000, k_neighbors=5)

# Unimos: Train Aumentado
df_train_augmented = pd.concat([df_train_real, df_sintetico], ignore_index=True)
print(f"🚀 Train Set Aumentado: {len(df_train_augmented)} muestras listas para entrenar.")

# ==============================================================================
# 3. DEFINICIÓN DE EXPERIMENTOS
# ==============================================================================

# A) SMALL
feats_small = [
    'Superficie', 'Localizacion_Clave', 'Num_baños', 'Dormitorios', 
    'Ascensor', 'Planta_ordinal', 'Año_de_construccion'
]

# B) MEDIUM
feats_medium = feats_small + [
    'Renta_Neta_Media', 'nivel_lujo', 'calidad_iluminacion', 
    'Garaje', 'Terraza_balcon', 'Superficie_x_Lujo', 
    'Servicios_cercanos', 'Estado_conservacion', 'Seguridad_portero'
]

# C) LARGE
exclude_cols = ['Precio', 'Precio_m2_Real', 'Log_Precio', 'Precio_m2', 
                'Descripción', 'Subjetividad_desc', 'Longitud_desc', 
                'Adjetivos_positivos', 'Urgencia_desc']
feats_large = [c for c in df.columns if c not in exclude_cols]

experiments = {
    'SMALL (Esencial)': feats_small,
    'MEDIUM (Smart)': feats_medium,
    'LARGE (Completo)': feats_large
}

# ==============================================================================
# 4. BUCLE DE ENTRENAMIENTO Y EVALUACIÓN
# ==============================================================================
results = []
print("\n🔥 INICIANDO TORNEO (Entrenando con Augmented -> Validando con Real)...\n")

for name, features in experiments.items():
    print(f"⚙️ Modelo: {name} | Vars: {len(features)}")
    
    # Filtramos columnas válidas
    valid_feats = [f for f in features if f in df_train_augmented.columns]
    
    # PREPARAR TRAIN (Aumentado)
    X_train = df_train_augmented[valid_feats]
    y_train = np.log1p(df_train_augmented['Precio']) # Log Target
    
    # PREPARAR TEST (Real Puro)
    X_test = df_test_real[valid_feats]
    y_test_real_euros = df_test_real['Precio'] # Para validar en euros
    
    # Detectar tipos de columnas dinámicamente
    num_cols = X_train.select_dtypes(include=['number']).columns.tolist()
    cat_cols = X_train.select_dtypes(include=['object', 'category', 'bool']).columns.tolist()
    
    # Pipeline
    transformers = [('num', Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler())]), num_cols)]
    
    if cat_cols:
        transformers.append(('cat', TargetEncoder(target_type='continuous', smooth=10.0), cat_cols))
        
    preprocessor = ColumnTransformer(transformers)
    
    # Modelo XGBoost Estándar
    model = XGBRegressor(
        n_estimators=1000, learning_rate=0.05, max_depth=5, 
        subsample=0.8, colsample_bytree=0.8, n_jobs=-1, random_state=SEED
    )
    
    pipeline = Pipeline([('preprocessor', preprocessor), ('model', model)])
    
    # Entrenar
    pipeline.fit(X_train, y_train)
    
    # Predecir
    pred_log = pipeline.predict(X_test)
    pred_euros = np.expm1(pred_log)
    
    # Métricas
    mae = mean_absolute_error(y_test_real_euros, pred_euros)
    mape = np.mean(np.abs((y_test_real_euros - pred_euros) / y_test_real_euros)) * 100
    
    print(f"   -> MAE Test: {mae:,.0f} € | MAPE: {mape:.2f}%\n")
    
    results.append({'Modelo': name, 'Num_Vars': len(valid_feats), 'MAE (€)': mae, 'MAPE (%)': mape})

# ==============================================================================
# 5. VISUALIZACIÓN
# ==============================================================================
res_df = pd.DataFrame(results).sort_values('MAE (€)')

print("="*60)
print("🏆 CLASIFICACIÓN FINAL")
print("="*60)
print(res_df.to_string(index=False, float_format=lambda x: f"{x:,.2f}" if isinstance(x, float) else x))

plt.figure(figsize=(10, 6))
sns.barplot(data=res_df, x='Modelo', y='MAE (€)', palette='viridis')
plt.title('Error Real en Test Set (Entrenado con Data Augmentation)', fontsize=14)
plt.ylabel('Error Medio Absoluto (€)')
plt.grid(axis='y', alpha=0.3)
for i, v in enumerate(res_df['MAE (€)']):
    plt.text(i, v + 500, f"{v:,.0f} €", ha='center', fontweight='bold')
plt.show()
```

```python
import pandas as pd
import numpy as np
import warnings
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder, TargetEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.metrics import mean_absolute_error
from xgboost import XGBRegressor

# Configuración
warnings.filterwarnings('ignore')
SEED = 42
np.random.seed(SEED)

# ==============================================================================
# 1. CARGA DE DATOS
# ==============================================================================
ruta_parquet = "C:/Users/mario/OneDrive/Mario/CARRERA/TFG/ANALIZADOR-TECNOCASA/DATOS/tecnocasa_modelo.parquet"
# Si falla la ruta absoluta, intenta con relativa:
# ruta_parquet = "DATOS/tecnocasa_modelo.parquet"

df = pd.read_parquet(ruta_parquet, engine='fastparquet')

# Filtro Outliers (Limpieza básica para todos)
q_low, q_high = df['Precio'].quantile(0.01), df['Precio'].quantile(0.99)
df = df[(df['Precio'] > q_low) & (df['Precio'] < q_high)].copy()

# ==============================================================================
# 2. DEFINICIÓN DE LOS 3 MODELOS (SELECCIÓN DE VARIABLES)
# ==============================================================================

# --- MODELO PEQUEÑO (SMALL): "Las Esenciales" ---
# Variables estructurales puras. Sin esto no hay tasación.
feats_small = [
    'Superficie',           # El Rey
    'Localizacion_Clave',   # La Reina (Barrio)
    'Num_baños',            # Más correlacionado que dormitorios
    'Dormitorios',          
    'Ascensor',             # Crítico en pisos antiguos
    'Planta_ordinal',       # Altura (Ático vs Bajo)
    'Año_de_construccion'   # Antigüedad
]

# --- MODELO MEDIANO (MEDIUM): "Smart & Clean" ---
# Small + Variables de Alta Calidad (IA + Contexto) + Extras Valiosos
feats_medium = feats_small + [
    'Renta_Neta_Media',      # Contexto socioeconómico (Poderoso)
    'nivel_lujo',            # IA Gemini (Estado visual)
    'calidad_iluminacion',   # IA Gemini (Luz)
    'Garaje',                # Extra de valor
    'Terraza_balcon',        # Extra de valor
    'Superficie_x_Lujo',     # Feature Engineering (Interacción)
    'Servicios_cercanos',    # Agregado de conectividad
    'Estado_conservacion',   # Importante si está disponible
    'Seguridad_portero'      # Valor añadido en zonas ricas
]

# --- MODELO GRANDE (LARGE): "Todo Incluido" ---
# Usamos todas las columnas excepto el target y textos crudos
exclude_cols = ['Precio', 'Precio_m2_Real', 'Log_Precio', 'Precio_m2', 
                'Descripción', 'Subjetividad_desc', 'Longitud_desc', 'Adjetivos_positivos', 'Urgencia_desc']
feats_large = [c for c in df.columns if c not in exclude_cols]

# Diccionario de experimentos
experiments = {
    'SMALL (Esencial)': feats_small,
    'MEDIUM (Smart)': feats_medium,
    'LARGE (Completo)': feats_large
}

# ==============================================================================
# 3. MOTOR DE ENTRENAMIENTO
# ==============================================================================
results = []

print(f"🚀 INICIANDO TORNEO DE MODELOS (N={len(df)} muestras)\n")

for name, features in experiments.items():
    print(f"⚙️ Entrenando: {name} | Variables: {len(features)}")
    
    # 1. Filtrar dataset
    # Aseguramos que solo usamos columnas que existen
    valid_features = [f for f in features if f in df.columns]
    X = df[valid_features]
    y = df['Precio']
    
    # 2. Split
    X_train, X_test, y_train, y_test = train_test_split(
        X, np.log1p(y), test_size=0.2, random_state=SEED
    )
    
    # 3. Preprocesador Automático
    # Detectamos tipos dinámicamente según el subset actual
    num_cols = X.select_dtypes(include=['number']).columns.tolist()
    cat_cols = X.select_dtypes(include=['object', 'category', 'bool']).columns.tolist()
    
    # Pipeline de transformaciones
    # IMPORTANTE: TargetEncoder para zonas en todos los modelos para ser justos
    transformers = [
        ('num', Pipeline([
            ('imputer', SimpleImputer(strategy='median')),
            ('scaler', StandardScaler())
        ]), num_cols)
    ]
    
    if cat_cols:
        # Usamos TargetEncoder para categóricas de alta cardinalidad (Barrios) 
        # y OneHot para las pequeñas si las hubiera, pero TargetEncoder es robusto para todo.
        transformers.append(
            ('cat', TargetEncoder(target_type='continuous', smooth=10.0), cat_cols)
        )
    
    preprocessor = ColumnTransformer(transformers)
    
    # 4. Modelo (XGBoost Robusto)
    # Usamos parámetros conservadores para evitar overfitting en el modelo grande
    model = XGBRegressor(
        n_estimators=1000,
        learning_rate=0.05,
        max_depth=5,
        subsample=0.8,
        colsample_bytree=0.8,
        n_jobs=-1,
        random_state=SEED
    )
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('model', model)
    ])
    
    # 5. Entrenar y Evaluar
    pipeline.fit(X_train, y_train)
    
    # Predicción (Inversa Log)
    preds_log = pipeline.predict(X_test)
    preds_euros = np.expm1(preds_log)
    y_test_euros = np.expm1(y_test)
    
    mae = mean_absolute_error(y_test_euros, preds_euros)
    mape = np.mean(np.abs((y_test_euros - preds_euros) / y_test_euros)) * 100
    
    print(f"   -> MAE: {mae:,.0f} € | MAPE: {mape:.2f}%\n")
    
    results.append({
        'Modelo': name,
        'Num_Vars': len(valid_features),
        'MAE (€)': mae,
        'MAPE (%)': mape
    })

# ==============================================================================
# 4. RESULTADOS Y GRÁFICA
# ==============================================================================
res_df = pd.DataFrame(results).sort_values('MAE (€)')

print("="*60)
print("🏆 RESULTADOS FINALES DEL EXPERIMENTO")
print("="*60)
print(res_df.to_string(index=False, float_format=lambda x: f"{x:,.2f}" if isinstance(x, float) else x))

# Visualización
plt.figure(figsize=(10, 6))
sns.barplot(data=res_df, x='Modelo', y='MAE (€)', palette='viridis')
plt.title('Evolución del Error según Complejidad del Modelo', fontsize=14)
plt.ylabel('Error Medio Absoluto (€)')
plt.grid(axis='y', alpha=0.3)
for i, v in enumerate(res_df['MAE (€)']):
    plt.text(i, v + 1000, f"{v:,.0f} €", ha='center', fontweight='bold')
plt.show()
```

### **VERSION LARGE**

```python
import pandas as pd
import numpy as np
import warnings
import joblib
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder, TargetEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.metrics import mean_absolute_error
from xgboost import XGBRegressor
from sklearn.neighbors import NearestNeighbors
from sklearn.base import BaseEstimator, RegressorMixin

# Configuración
warnings.filterwarnings('ignore')
SEED = 42
np.random.seed(SEED)

# ==============================================================================
# 1. CLASE HÍBRIDA (XGBoost + Corrección KNN)
# ==============================================================================
class HybridValuator(BaseEstimator, RegressorMixin):
    def __init__(self, base_model, n_neighbors=5, correction_weight=0.5):
        """
        base_model: El modelo XGBoost/Stacking optimizado.
        n_neighbors: Cuántos testigos mirar para corregir el error.
        correction_weight: Cuánto caso hacerle a la corrección (0.0 a 1.0).
        """
        self.base_model = base_model
        self.n_neighbors = n_neighbors
        self.correction_weight = correction_weight
        self.knn = NearestNeighbors(n_neighbors=n_neighbors)
        self.X_train_processed = None
        self.y_train_ratios = None # Ratio Real / Predicho
        
    def fit(self, X, y):
        # 1. Entrenar modelo base
        self.base_model.fit(X, y)
        
        # 2. Predecir sobre el propio Train para ver dónde falla
        preds_log = self.base_model.predict(X)
        preds_euros = np.expm1(preds_log)
        y_euros = np.expm1(y)
        
        # 3. Calcular el Ratio de Error (Sesgo Local)
        # Si Ratio > 1, el modelo subestimó (Real > Pred). 
        # Si Ratio < 1, el modelo sobreestimó.
        self.y_train_ratios = y_euros / preds_euros
        
        # 4. Entrenar KNN sobre las features procesadas (Extraemos del pipeline interno si es posible)
        # Truco: Usamos el preprocesador del pipeline base para transformar X
        try:
            self.preprocessor = self.base_model.named_steps['preprocessor']
            self.X_train_processed = self.preprocessor.transform(X)
        except:
            # Si no es un pipeline, asumimos X ya procesada (menos robusto)
            self.X_train_processed = X
            
        self.knn.fit(self.X_train_processed)
        return self
    
    def predict(self, X):
        # 1. Predicción Base (Conservadora)
        base_log = self.base_model.predict(X)
        base_euros = np.expm1(base_log)
        
        # 2. Buscar vecinos para corrección
        X_processed = self.preprocessor.transform(X)
        distances, indices = self.knn.kneighbors(X_processed)
        
        # 3. Calcular Factor de Corrección Local
        # Promedio del ratio de error de los vecinos
        corrections = []
        for idx_row in indices:
            # Media de los ratios de los vecinos
            neighbor_ratios = self.y_train_ratios.iloc[idx_row] if hasattr(self.y_train_ratios, 'iloc') else self.y_train_ratios[idx_row]
            # Usamos mediana para evitar outliers en los vecinos
            corrections.append(np.median(neighbor_ratios))
            
        correction_factors = np.array(corrections)
        
        # 4. Aplicar Corrección Suavizada
        # Predicción Final = Predicción Base * (1 + weight * (Factor - 1))
        # Si weight=1, aplicamos toda la corrección. Si weight=0.5, la mitad.
        final_preds = base_euros * (1 + self.correction_weight * (correction_factors - 1))
        
        # Retornamos en LOG para ser consistentes con scikit-learn scores, 
        # o Euros directo. Aquí devolveremos Euros para evaluar fuera.
        return np.log1p(final_preds)

# ==============================================================================
# 2. CARGA DE DATOS (MODELO LARGE REAL)
# ==============================================================================
ruta_parquet = "C:/Users/mario/OneDrive/Mario/CARRERA/TFG/ANALIZADOR-TECNOCASA/DATOS/tecnocasa_modelo.parquet"
df = pd.read_parquet(ruta_parquet, engine='fastparquet')

# Filtro Outliers
q_low, q_high = df['Precio'].quantile(0.01), df['Precio'].quantile(0.99)
df = df[(df['Precio'] > q_low) & (df['Precio'] < q_high)].copy()

# Variables Large
if 'Localizacion_Clave' in df.columns:
    enc = TargetEncoder(smooth=10.0, target_type='continuous')
    df['Barrio_Value'] = enc.fit_transform(df[['Localizacion_Clave']], df['Precio'])
    df.drop(columns=['Localizacion_Clave'], inplace=True)

exclude_cols = ['Precio', 'Precio_m2_Real', 'Log_Precio', 'Precio_m2', 
                'Descripción', 'Subjetividad_desc', 'Longitud_desc', 
                'Adjetivos_positivos', 'Urgencia_desc', 'Desc_lemmatized']
features = [c for c in df.columns if c not in exclude_cols]

X = df[features]
y = df['Precio']
y_log = np.log1p(y)

# Split
X_train, X_test, y_train_log, y_test_log, y_train_real, y_test_real = train_test_split(
    X, y_log, y, test_size=0.2, random_state=SEED
)

# ==============================================================================
# 3. PIPELINE BASE
# ==============================================================================
num_cols = X.select_dtypes(include=['number']).columns
cat_cols = X.select_dtypes(include=['object', 'category', 'bool']).columns

preprocessor = ColumnTransformer([
    ('num', Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler())]), num_cols),
    ('cat', Pipeline([('imputer', SimpleImputer(strategy='most_frequent')), ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))]), cat_cols)
])

# XGBoost Base (Tu mejor modelo hasta ahora)
xgb_base = XGBRegressor(
    n_estimators=1500, learning_rate=0.03, max_depth=6, 
    subsample=0.8, colsample_bytree=0.8, n_jobs=-1, random_state=SEED
)

pipeline_base = Pipeline([
    ('preprocessor', preprocessor),
    ('model', xgb_base)
])

# ==============================================================================
# 4. ENTRENAMIENTO HÍBRIDO
# ==============================================================================
print("\n🚀 Entrenando Modelo Híbrido (XGBoost + Corrección de Vecinos)...")

# Instanciamos el Híbrido
# weight=0.7 significa "Confía un 70% en la corrección de los vecinos y un 30% en el modelo original"
hybrid_model = HybridValuator(base_model=pipeline_base, n_neighbors=5, correction_weight=0.7)

hybrid_model.fit(X_train, y_train_log)

# ==============================================================================
# 5. EVALUACIÓN Y DESGLOSE
# ==============================================================================
pred_log = hybrid_model.predict(X_test)
pred_euros = np.expm1(pred_log)

mae_global = mean_absolute_error(y_test_real, pred_euros)
mape_global = np.mean(np.abs((y_test_real - pred_euros) / y_test_real)) * 100

print("="*50)
print(f"RESULTADO FINAL (HÍBRIDO)")
print("="*50)
print(f"📉 MAE Global: {mae_global:,.0f} €")
print(f"📉 MAPE Global: {mape_global:.2f}%")

# Desglose por Decil
df_eval = pd.DataFrame({'Real': y_test_real, 'Pred': pred_euros})
df_eval['Error_Abs'] = np.abs(df_eval['Real'] - df_eval['Pred'])
df_eval['Error_Pct'] = (df_eval['Error_Abs'] / df_eval['Real']) * 100
df_eval['Intervalo'] = pd.qcut(df_eval['Real'], 10, duplicates='drop')

metrics = df_eval.groupby('Intervalo', observed=False)[['Error_Abs', 'Error_Pct']].mean()

print("\n📊 Nuevo Desglose:")
pd.options.display.float_format = '{:,.2f}'.format
print(metrics)

```
