# 🏘️ Preparación y Selección de Variables para Predicción de Precios Inmobiliarios - REDPISO

Este notebook tiene como objetivo preparar un conjunto de datos de viviendas extraído de Redpiso para entrenar modelos de predicción de precios. El flujo completo incluye la carga y limpieza de datos. A continuación, se describen los principales bloques del notebook:

#### 🔹 Limpieza de variables
Se eliminan columnas irrelevantes o redundantes, como identificadores únicos o columnas derivadas del objetivo (`Precio_m2`). Se eliminan columnas irrelevantes o redundantes, así como registros con precios nulos o no válidos. Además, se excluyen barrios con pocos inmuebles para garantizar representatividad.

#### 🔹 Filtrado por varianza
Se eliminan variables con baja varianza, tanto numéricas como categóricas, ya que aportan escasa información al modelo.

#### 🔹 Selección de características relevantes
Se emplean técnicas estadísticas como:
- `SelectKBest` con `f_regression` para las numéricas.
- `chi2` para las categóricas discretizadas según el precio.
  
También se analiza la multicolinealidad entre variables numéricas.

#### 🔹 Eliminación de outliers
Se filtran valores atípicos mediante el rango intercuartílico (IQR) y el modelo `Isolation Forest`, eliminando observaciones con precios o superficies fuera de escala.

#### 🔹 Creación de nuevas variables
Se generan variables derivadas que capturan relaciones relevantes, como ratios de superficie por dormitorios o baños, edad del inmueble, nivel de lujo y cantidad de servicios cercanos. Estas ayudan al modelo a detectar patrones complejos del mercado inmobiliario.


Este preprocesamiento permite disponer de un conjunto limpio, reducido y óptimo para entrenar modelos predictivos con mayor precisión y menor riesgo de overfitting.

---

### 🔧 Importación de librerías y carga de datos
Importamos las principales librerías de análisis, visualización y preprocesamiento. A continuación, cargamos el dataset en formato parquet para su posterior limpieza y análisis.

```python
# Importaciones y configuración
import pandas as pd
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, LabelEncoder
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.feature_selection import SelectKBest, f_regression, chi2, VarianceThreshold
from sklearn.ensemble import IsolationForest
import warnings
warnings.filterwarnings('ignore')
```

```python
# ==============================================================================
# 1. CARGA DEL DATASET
# ==============================================================================
# Ajusta la ruta a tu parquet o csv
df_final = pd.read_parquet("C:/Users/mario/TFG/ANALIZADOR-TECNOCASA/DATOS/redpiso_RentaNetaMedia.parquet", engine='fastparquet')
```

### 📑 Clasificación de variables por tipo
Identificamos las columnas numéricas, binarias (0/1) y categóricas para poder aplicarles el tratamiento adecuado en etapas posteriores del preprocesamiento.



```python
# ==============================================================================
# 2. HOMOGENEIZACIÓN DE COLUMNAS (Para adaptar al código original)
# ==============================================================================
renombres = {
    'precio_num': 'Precio',
    'superficie_m2': 'Superficie',
    'habitaciones': 'Dormitorios',
    'banos': 'Num_baños',
    'anio_construccion': 'Año_de_construccion',
    'estado_detallado': 'Estado_conservacion',
    'ascensor': 'Ascensor',
    'aire_acondicionado': 'Aire_acondicionado',
    'parking': 'Garaje',
    'trastero': 'Trastero',
    'jardin': 'Jardin',
    'piscina': 'Piscina',
    'amueblado': 'Amueblado'
}
df_final = df_final.rename(columns=renombres)
```

```python
# ==============================================================================
# 3. TRATAMIENTO INTELIGENTE DE NULOS (Vital para RedPiso)
# ==============================================================================
print(f"Shape inicial: {df_final.shape}")

# 3.1. Eliminar filas sin Precio o Superficie
df_final = df_final.dropna(subset=['Precio', 'Superficie'])

# 3.2. Rellenar Metros Útiles (Estimación: 85% de la superficie construida)
df_final['metros_utiles'] = df_final['metros_utiles'].fillna(df_final['Superficie'] * 0.85)

# 3.3. Rellenar variables numéricas de cantidad con 0
cols_fill_zero = ['numero_terrazas', 'garajes']
for col in cols_fill_zero:
    df_final[col] = df_final[col].fillna(0)

# 3.4. Rellenar numéricas básicas con la Mediana
cols_fill_median = ['Dormitorios', 'Num_baños', 'Año_de_construccion', 'telefono']
for col in cols_fill_median:
    df_final[col] = df_final[col].fillna(df_final[col].median())

# 3.5. Rellenar categóricas con "Desconocido"
cols_fill_desc = [
    'calificacion_energetica', 'Estado_conservacion', 'calefaccion', 
    'tipo_suelo', 'antiguedad', 'agua_caliente', 'orientacion', 'tipo_fachada'
]
for col in cols_fill_desc:
    df_final[col] = df_final[col].fillna('Desconocido')
```

### 🧹 Limpieza de columnas irrelevantes o redundantes
Eliminamos variables que no aportan valor predictivo (como enlaces, IDs, columnas derivadas o redundantes con la localización).

```python
# ==============================================================================
# 4. LIMPIEZA DE COLUMNAS SIN VALOR
# ==============================================================================
columns_to_drop = [
    'url', 'titulo', 'referencia', 'precio', 'estado', 'oficina', 'telefono',
    'fecha_extraccion', 'visitado_fecha', 'datos_completos', 'Direccion_Extraida',
    'precio_por_m2' # IMPORTANTÍSIMO ELIMINAR ESTA (Data leakage)
]
df_final = df_final.drop(columns=columns_to_drop, errors='ignore')

# Convertir booleanas a 1/0
cols_bool = [
    'Ascensor', 'Aire_acondicionado', 'Garaje', 'Trastero', 'tiene_terraza', 
    'balcon', 'Jardin', 'Piscina', 'Amueblado', 'cocina_equipada', 'tour_virtual'
]
for col in cols_bool:
    if col in df_final.columns:
        df_final[col] = df_final[col].astype(int)
```

```python
# ==============================================================================
# 5. DEFINICIÓN DE VARIABLES
# ==============================================================================
numericas = [
    'Precio', 'Dormitorios', 'Superficie', 'Num_baños', 'Año_de_construccion',
    'metros_utiles', 'numero_terrazas', 'garajes', 'Renta_Neta_Media'
]

binarias = cols_bool

categoricas = [
    'Estado_conservacion', 'calificacion_energetica', 'calefaccion', 
    'tipo_suelo', 'antiguedad', 'agua_caliente', 'orientacion', 
    'tipo_fachada', 'Localizacion_Clave'
]

vars_validas = numericas + binarias
```

```python
# Actualizar variables
numericas = [col for col in numericas if col in df_final.columns]
binarias = [col for col in binarias if col in df_final.columns]
categoricas = [col for col in categoricas if col in df_final.columns]
```

### 🚫 Eliminación de variables con baja varianza
Descartamos variables numéricas, binarias y categóricas que presentan muy poca variabilidad entre observaciones, ya que probablemente no aporten valor en la predicción.

```python
vars_validas = [col for col in numericas + binarias if col in df_final.columns]
selector = VarianceThreshold(threshold=0.05)
selector.fit(df_final[vars_validas])

low_variance_num = [col for col in vars_validas
                   if col not in df_final[vars_validas].columns[selector.get_support()]]

print("Numéricas/Binarias con baja varianza:", low_variance_num)
```

```python
low_variance_cat = []
for col in categoricas:
    if col in df_final.columns:
        top_value_ratio = df_final[col].value_counts(normalize=True).iloc[0]
        if top_value_ratio > 0.95:
            low_variance_cat.append(col)
print("Categóricas con baja varianza:", low_variance_cat)
```

```python
columns_to_drop = list(set(low_variance_num + low_variance_cat))
columns_to_drop = [col for col in columns_to_drop if col in df_final.columns]
df_final = df_final.drop(columns=columns_to_drop)

print(f"Se eliminaron {len(columns_to_drop)} columnas: {columns_to_drop}")
print(f"Nuevo shape: {df_final.shape}")
```

```python
# Actualizar variables
numericas = [col for col in numericas if col in df_final.columns]
binarias = [col for col in binarias if col in df_final.columns]
categoricas = [col for col in categoricas if col in df_final.columns]
```

### 🧮 Feature Selection (variables que más influyen en el Precio)
Aplicamos SelectKBest con estadística f_regression para identificar las 15 variables numéricas más relevantes en la predicción del precio.

```python
numerical_features = [col for col in numericas if col != 'Precio']

selector = SelectKBest(score_func=f_regression, k=15)  # Top 15 features
X_selected = selector.fit_transform(df_final[numerical_features], df_final['Precio'])

# Obtener nombres de las features seleccionadas
selected_num_features = df_final[numerical_features].columns[selector.get_support()]
scores = selector.scores_[selector.get_support()]

feature_ranking = pd.DataFrame({
    'Feature': selected_num_features,
    'F_Score': scores
}).sort_values('F_Score', ascending=False)

print("Variables numéricas más informativas:", selected_num_features)
print(feature_ranking)
```

Convertimos el precio a una variable discreta (Precio_bin) para poder aplicar la prueba de independencia chi2, seleccionando las 10 variables categóricas con mayor poder explicativo.

```python
# Paso 1: Discretizar el precio
df_final['Precio_bin'] = pd.qcut(df_final['Precio'], q=4, labels=False)

# Paso 2: Codificar las categóricas
X_cat = df_final[categoricas].copy()
for col in X_cat.columns:
    X_cat[col] = X_cat[col].astype(str)
    X_cat[col] = LabelEncoder().fit_transform(X_cat[col])

# Paso 3: Chi2
selector = SelectKBest(score_func=chi2, k=10)
selector.fit(X_cat, df_final['Precio_bin'])

selected_cat_features = list(X_cat.columns[selector.get_support()])
df_final = df_final.drop(columns = "Precio_bin")
print("Variables categóricas más informativas:", selected_cat_features)
```

### 🧩 Detección de multicolinealidad entre variables numéricas
Calculamos la matriz de correlaciones para eliminar variables altamente correlacionadas entre sí (r > 0.9), lo cual puede afectar negativamente al rendimiento y la interpretabilidad del modelo.

```python
corr_matrix = df_final[numericas].corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
high_corr = [column for column in upper.columns if any(upper[column] > 0.9)]

print("Variables numéricas muy correlacionadas:", high_corr)
```

```python
to_drop = set()

for col in high_corr:
    corr_targets = upper[col][upper[col] > 0.9].index.tolist()
    
    # Conservar la variable con mayor correlación con el precio
    if col in df_final.columns:
        best_feature = max([col] + corr_targets, 
                          key=lambda x: abs(df_final[x].corr(df_final['Precio'])))
        
        to_drop.update([f for f in corr_targets + [col] if f != best_feature])

print("\nVariables a eliminar (redundantes):", to_drop)
```

```python
df_final = df_final.drop(columns=[col for col in to_drop if col in df_final.columns])
numericas = [col for col in numericas if col in df_final.columns]
remaining_corr = df_final[numericas].corr().abs()
```

### 🏙️ Filtrado de barrios con pocos inmuebles
Se eliminan los barrios con menos de 4 viviendas registradas para garantizar que cada zona tenga suficiente representatividad estadística en el modelo.

### 🧹 Eliminación de outliers mediante IQR e Isolation Forest  
Se aplican dos métodos complementarios para detectar y eliminar valores atípicos en variables clave como precio, superficie, dormitorios y baños.  
El enfoque combina el rango intercuartílico (IQR) con el algoritmo *Isolation Forest* para identificar propiedades con valores anómalos.

```python
# ==============================================================================
# 9. GESTIÓN DE OUTLIERS
# ==============================================================================
# 9.1 Zonas poco representadas
conteo_barrios = df_final['Localizacion_Clave'].value_counts()
barrios_validos = conteo_barrios[conteo_barrios >= 4].index
df_final = df_final[df_final['Localizacion_Clave'].isin(barrios_validos)]

# 9.2 IQR
num_cols_outliers = ['Precio', 'metros_utiles']
def remove_outliers_iqr(df, cols, factor=2.0):
    for col in cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        mask = (df[col] >= Q1 - factor * IQR) & (df[col] <= Q3 + factor * IQR)
        df = df[mask]
    return df

df_final = remove_outliers_iqr(df_final, num_cols_outliers, factor=2.0)

# 9.3 Isolation Forest
iso = IsolationForest(contamination=0.02, random_state=42)
mask_iso = iso.fit_predict(df_final[num_cols_outliers]) == 1
df_final = df_final[mask_iso]
print(f"\nShape final tras limpieza de outliers: {df_final.shape}")
```

### 🧮 Feature Engineering: creación de variables derivadas
Se generan nuevas características que ayudan al modelo a capturar relaciones no lineales entre variables.  
Estas incluyen ratios de espacio por habitación, edad del inmueble, nivel de lujo ponderado y cantidad de servicios cercanos.

```python
# ==============================================================================
# 10. FEATURE ENGINEERING (Nuevas variables)
# ==============================================================================
df_final['Dormitorios_por_m2'] = df_final['Dormitorios'] / df_final['metros_utiles']
df_final['Banos_por_m2'] = df_final['Num_baños'] / df_final['metros_utiles']
df_final['Recien_reformado'] = (df_final['Estado_conservacion'] == 'Perfecto Estado').astype(int)
df_final['Edad_inmueble'] = 2026 - df_final['Año_de_construccion']
```
